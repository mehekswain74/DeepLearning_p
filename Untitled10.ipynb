```python
import tensorflow
from tensorflow import keras
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense,Flatten
```

```python
(x_train,y_train),(x_test,y_test)=keras.datasets.mnist.load_data()
```

```python
x_train.shape
```

```python
import matplotlib.pyplot as plt
plt.imshow(x_train[1])
```

```python
'''As the values in this data set ranges from 0 to 255 so we need to make
   the values in similar range so that our neural network converge faster'''
```

```python
x_train[0]
```

```python
# dividing the pixels value with 255 to make it in a rage from 0-1
# if the value is 255 div with it will give 1 anf if 0 then 0
x_train = x_train/255
x_test = x_test/255
```

```python
x_train[0]
```

```python
# As the dataset is in 28X28 format we need sequentially arrange in 1 row
# we are using Flatten(converts higer dimention in 1D) for it
model = Sequential()
model.add(Flatten(input_shape=(28,28)))
model.add(Dense(128,activation='relu'))
model.add(Dense(38,activation='relu'))  # for increasing accuracy of model
model.add(Dense(10,activation='softmax'))
```

```python
model.summary()
```

```python
# by using sparse_categorical_crossentropy we dont have to convert the
# data(labels) into one hot encoding
model.compile(loss='sparse_categorical_crossentropy',optimizer='Adam',metrics=['Accuracy'])
```

```python
# training our model
history=model.fit(x_train,y_train,epochs=25,validation_split=0.2)
```

```python
# pridiction
model.predict(x_test)
```

```python
y_prob = model.predict(x_test)
```

```python
# we need to show in output weather the image is 1,2 or any other no
# to do so in every prob array we nedd to finf in which index the value is highest
y_pred = y_prob.argmax(axis=1)
```

```python
y_pred
```

```python
# checking accuracy score
from sklearn.metrics import accuracy_score
accuracy_score(y_test,y_pred)
```

```python
# how to improve accuracy there are various ways...
# changing arckhitecture like -> add more layers
#                             -> increasing existing nodes in layers
#                             -> insreasing the epochs
```

```python
plt.plot(history.history['loss'],label='train_loss')
plt.plot(history.history['val_loss'],label='val_loss')
plt.legend()
```

```python
plt.plot(history.history['Accuracy'],label='accuracy')
plt.plot(history.history['val_Accuracy'],label='valaccuracy')
plt.legend()
```
